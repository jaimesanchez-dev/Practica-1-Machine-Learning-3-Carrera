# Carga del modelo final y realización de las predicciones sobre los datos de competición

## 1. Carga del modelo

```python
import pandas as pd
import pickle as pkl
from sklearn.preprocessing import LabelEncoder

# Cargamos el modelo final
with open("final_model.pkl", "rb") as f:
    model = pkl.load(f)

print("Modelo final cargado correctamente.")
```

## 2. Carga de los datos

```python
try:
    df_competicion = pd.read_pickle("dataset/bank_competition_10.pkl")
    print(f"Datos de competición cargados: {df_competicion.shape[0]} registros.")
except FileNotFoundError:
    print("Error: No se encuentra el archivo 'bank_competition_10.pkl'.")
```

## 3. Realización de las predicciones

```python
predictions = model.predict(df_competicion)

print("Predicciones generadas con éxito.")
```

## 4. Creación de los archivos de entrega


```python
df_entrega = pd.DataFrame(predictions, columns=['prediction'])

# Guardar a CSV
df_entrega.to_csv("predicciones.csv", index=False)

print("Archivo 'predicciones.csv' generado con éxito.")
```
